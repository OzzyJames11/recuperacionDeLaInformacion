## Parte 1: Generación de Embeddings

Vamos a utilizar E5 como modelo de embeddings.

La documentación de E5 está disponible desde este [link](https://huggingface.co/intfloat/e5-base-v2)

### Actividad

1. Normalizar el corpus
2. Definir una función `chunk_text`, y dividir los textos en _chunks_.
3. Generar embeddings por cada _chunk_

```python
import pandas as pd
import numpy as np
from tqdm.auto import tqdm
import re

df = df.dropna(subset=["text"]).reset_index(drop=True)

# Limpieza básica
def normalize_text(s: str) -> str:
    s = re.sub(r"\s+", " ", s).strip()
    return s

df["text_norm"] = df["text"].astype(str).map(normalize_text)

df.head()
```

```python
def chunk_text(text: str, max_chars: int = 800, overlap: int = 100):
    """
    Chunking por caracteres.
    max_chars ~ 600-1000 suele funcionar bien.
    overlap ayuda a no cortar ideas a la mitad.
    """
    chunks = []
    start = 0
    n = len(text)
    while start < n:
        end = min(start + max_chars, n)
        chunk = text[start:end]
        chunk = chunk.strip()
        if len(chunk) > 0:
            chunks.append(chunk)
        if end == n:
            break
        start = max(0, end - overlap)
    return chunks

records = []
for i, row in df.iterrows():
    chunks = chunk_text(row["text_norm"], max_chars=800, overlap=100)
    for j, ch in enumerate(chunks):
        records.append({
            "doc_id": int(i),
            "chunk_id": j,
            "text": ch
        })

chunks_df = pd.DataFrame(records)
chunks_df.head(), len(chunks_df)
```

```python
from sentence_transformers import SentenceTransformer

MODEL_NAME = "intfloat/e5-base-v2"   # recomendado para retrieval
model = SentenceTransformer(MODEL_NAME)

# Textos a indexar (pasajes)
passages = ["passage: " + t for t in chunks_df["text"].tolist()]
```

```python
# Embeddings (N x D)
# Se debe usar normalize_embeddings=True para similitud coseno
embeddings = model.encode(
    passages,
    batch_size=16,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
).astype("float32")
```

```python
print(embeddings.shape, embeddings.dtype)
```

```python
def embed_query(query: str) -> np.ndarray:
    q = "query: " + query
    vec = model.encode(
        [q],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")
    return vec

query_text = "Battery measuring"

query_vec = embed_query(query_text)
query_vec.shape
```

## Parte 2: FAISS

FAISS es una librería para búsqueda por similitud eficiente y clustering de vectores densos.

La documentación de FAISS está disponible en este [link](https://faiss.ai/index.html)

### Actividad

```python
import faiss
import numpy as np

# Dimensión del embedding
D = embeddings.shape[1]

# Índice exacto con producto interno (cosine)
index = faiss.IndexFlatIP(D)

# Agregar embeddings
index.add(embeddings)

print("Vectores indexados:", index.ntotal)
```

```python
k = 10
scores, indices = index.search(query_vec, k)

for rank, idx in enumerate(indices[0]):
    print(f"Rank {rank+1} | Score: {scores[0][rank]:.4f}")
    print(chunks_df.iloc[idx]["text"][:300])
    print("-" * 80)
```

```python
from pymilvus import connections

connections.connect(
    alias="default",
    uri="milvus_demo.db"  # archivo local
)
```

```python
# Definir esquema
from pymilvus import FieldSchema, CollectionSchema, DataType, Collection
```

```python
DIM = embeddings.shape[1]

fields = [
    FieldSchema(name="id", dtype=DataType.INT64, is_primary=True),
    FieldSchema(name="embedding", dtype=DataType.FLOAT_VECTOR, dim=DIM),
    FieldSchema(name="text", dtype=DataType.VARCHAR, max_length=2048)
]

schema = CollectionSchema(fields, description="Document embeddings")
collection = Collection(name="docs_milvus", schema=schema)
```

```python
BATCH_SIZE = 500  # seguro para Milvus Lite
```

```python
from tqdm import tqdm

total = len(embeddings)

for start in tqdm(range(0, total, BATCH_SIZE)):
    end = min(start + BATCH_SIZE, total)

    batch_ids = list(range(start, end))
    batch_embeddings = embeddings[start:end].tolist()
    batch_texts = chunks_df["text"].iloc[start:end].tolist()

    collection.insert([
        batch_ids,
        batch_embeddings,
        batch_texts
    ])
```

```python
print("Número de entidades:", collection.num_entities)
```

```python
index_params_flat = {
    "metric_type": "COSINE",
    "index_type": "FLAT"
}

collection.create_index(
    field_name="embedding",
    index_params=index_params_flat
)
```

```python
collection.load()
```

```python
def milvus_search_flat(query_embedding, k):
    results = collection.search(
        data=query_embedding.tolist(),
        anns_field="embedding",
        param={"metric_type": "COSINE"},
        limit=k,
        output_fields=["text"]
    )

    return [
        (hit.id, hit.score, hit.entity.get("text"))
        for hit in results[0]
    ]
```

```python
results_flat = milvus_search_flat(query_vec, k=5)

for r in results_flat:
    print("Score:", r[1])
    print("Text:", r[2][:300])
    print("-" * 80)
```

```python
index_params_ivf = {
    "metric_type": "COSINE",
    "index_type": "IVF_FLAT",
    "params": {
        "nlist": 128
    }
}

collection.create_index(
    field_name="embedding",
    index_params=index_params_ivf
)
```

```python
collection.load()
```

```python
def milvus_search_ivf(query_embedding, k):
    results = collection.search(
        data=query_embedding.tolist(),
        anns_field="embedding",
        param={
            "metric_type": "COSINE",
            "params": {"nprobe": 8}
        },
        limit=k,
        output_fields=["text"]
    )

    return [
        (hit.id, hit.score, hit.entity.get("text"))
        for hit in results[0]
    ]
```

```python
results_ivf = milvus_search_ivf(query_vec, k=5)

for r in results_ivf:
    print("Score:", r[1])
    print("Text:", r[2][:300])
    print("-" * 80)
```

```python
# Mini Experimento
ids_flat = {r[0] for r in results_flat}
ids_ivf = {r[0] for r in results_ivf}

print("Overlap:", len(ids_flat & ids_ivf), "/ 5")
```

#### Respuestas a las Preguntas
- ¿Qué diferencia conceptual encuentras entre “schema + objetos” vs “tabla + filas”?
  
  Al tomar en consideración la primera opción, el schema define clases (colecciones) con propiedades tipadas en donde, cada registro es un objeto semántico (no solo una fila). También, los objetos pueden tener vectores explícitos, relaciones y un comportamiento semántico. Por otro lado, en el modelo relacional clásico (tablas + filas) el esquema define columnas fijas, en donde cada fila es un registro plano. Si existen relaciones, pero son externas (se hace uso de JOIN). Este modelo clásico es estructural, no semántico.
- ¿Cómo describirías el trade-off de complejidad vs expresividad?

  Al analizar varios aspectos como la curva de aprendizaje, se nota que Weavite es más complejo que SQL/tablas. También su infraestructura es más compleja y su expresividad semántica es muy alta, a diferencia de una expresividad semántica baja de SQL/tablas. La flexibilidad es alta en Weavite y rígida en sql. En conclusión, al usar objetos (Weavite) se obtiene expresividad semántica y capacidades avanzadas, sin embargo, se agrega una complejidad técnica mayor.

```python
import chromadb

chroma_client = chromadb.Client()

collection = chroma_client.create_collection(
    name="documents"
)
```

```python
import math

# Tamaño seguro de batch
BATCH_SIZE = 1000

ids = [str(i) for i in range(len(embeddings))]
docs = chunks_df["text"].tolist()
embs = embeddings.tolist()

num_batches = math.ceil(len(embs) / BATCH_SIZE)

for i in range(num_batches):
    start = i * BATCH_SIZE
    end = start + BATCH_SIZE

    collection.add(
        ids=ids[start:end],
        embeddings=embs[start:end],
        documents=docs[start:end]
    )

print(f"Inserción completada: {len(embs)} documentos")
```

```python
# Consulta Top-k
def chroma_search(query_embedding, k=5):
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=k
    )

    return [
        {
            "id": results["ids"][0][i],
            "text": results["documents"][0][i],
            "distance": results["distances"][0][i]
        }
        for i in range(k)
    ]
```

```python
from sentence_transformers import SentenceTransformer

# Se vuelve a cargar el modelo
MODEL_NAME = "intfloat/e5-base-v2"
embedder = SentenceTransformer(MODEL_NAME)
```

```python
# Se construye la query correctamente
query_text = "query: neural networks for image classification"
query_embedding = embedder.encode(query_text).tolist()
```

```python
# Ejemplo de consulta para un k=5
results = chroma_search(query_embedding, k=5)

for r in results:
    print(r["distance"], r["text"][:120])
```

#### Respuestas a las Preguntas
- ¿Qué tan fácil fue implementar todo comparado con Qdrant/Milvus?

- ¿Qué limitaciones ves para un sistema en producción?

    La escalabilidad limitada, se tiene un control pobre sobre índices, no hay persistencia básica y no es ideal para millones de vectores. Así, este es perfecto para pruebas y demos, pero no para producción grande ni sistemas distribuidos.

## Parte 7 — SQL + vectores: PostgreSQL/pgvector (vector search transparente)

### Objetivo
Guardar embeddings en una tabla y ejecutar una consulta SQL de similitud.

### Qué debes implementar
1. Conectar a una base PostgreSQL con `pgvector` habilitado.
2. Crear una tabla (ej. `documents`) con:
   - `id` (PK)
   - `text` (texto)
   - `embedding` (vector(D))
   - metadata (columnas adicionales)
3. Insertar todos los documentos y embeddings.
4. Consultar Top-k por similitud, ordenando por distancia.

### Fórmula conceptual (lo que implementa tu SQL)
Para una consulta `q`, buscas:
$$ argmin_d \in D \; \text{dist}(\vec{q}, \vec{d})$$
donde `dist` puede ser L2 o una variante para cosine (según configuración).

### Entregable
- Función `pgvector_search(query_embedding, k)` que ejecute SQL y devuelva:
  - id, score/distancia, text, metadata

### Preguntas
- ¿Qué tan “explicable” te parece esta aproximación vs las otras?
- ¿Qué ventajas ofrece el mundo SQL (JOIN, filtros, agregaciones)?
- ¿Qué limitaciones esperas en escalabilidad frente a bases vectoriales dedicadas?


```python
import numpy as np

def pgvector_like_search(query_emb, embeddings, texts, k=5):
    query = np.array(query_emb)
    embs = np.array(embeddings)

    # cosine similarity
    query = query / np.linalg.norm(query)
    embs = embs / np.linalg.norm(embs, axis=1, keepdims=True)

    scores = embs @ query
    topk_idx = np.argsort(scores)[-k:][::-1]

    return [
        {
            "rank": i+1,
            "score": float(scores[idx]),
            "text": texts[idx]
        }
        for i, idx in enumerate(topk_idx)
    ]
```

```python
# Se simula una tabla de SQL con un DF
import pandas as pd

documents_df = pd.DataFrame({
    "id": range(len(embeddings)),
    "text": chunks_df["text"].tolist(),
    "embedding": list(embeddings),
    "source": chunks_df.get("source", "unknown")
})
```

```python
def pgvector_search(query_embedding, k=5):
    results = pgvector_like_search(
        query_embedding,
        documents_df["embedding"].tolist(),
        documents_df["text"].tolist(),
        k
    )
    return results
```

```python
# Se simula una consulta
query_text = "neural networks for image classification"
query_embedding = model.encode("query: " + query_text)

results = pgvector_search(query_embedding, k=5)

for r in results:
    print(f"{r['rank']} | score={r['score']:.4f}")
    print(r["text"][:150])
    print("-" * 60)
```

#### Respuestas a las Preguntas
- ¿Qué tan “explicable” te parece esta aproximación vs las otras?

  Esta es la más entendible y explicable. Debido a que se ve explícitamente la columna en la que se guardan los embeddings, cómo se calcula la similitud y cómo se ordenan los resultados.
- ¿Qué ventajas ofrece el mundo SQL (JOIN, filtros, agregaciones)?

    JOIN: SQL tiene ventaja en este aspecto debido a que permite combinar búsqueda semántica con datos relacionales. Dado que en bases vectoriales puras esto es poco natural.
  
  Filtros: SQL tiene la capacidad de permitir filtros complejos, condiciones que pueden mezclar rangos, categorías, fechas y combinacines lógicas. Mientras que en bases vectoriales, si bien si existen los filtros, son más limitados.

  Agregaciones: SQL hace posible responder preguntas analíticas como cuántos documentos similares por categoría, qué tema aparece más en los tok-k, etc.
- ¿Qué limitaciones esperas en escalabilidad frente a bases vectoriales dedicadas?

    SQL pierde en escalabilidad debido a sus limitaciones. Comparar vectores suele ser una tarea complicada y se lo realiza con fuerza bruta $O(n)$, o con índices limitados. Su rendimiento suele caer a partir de millones de vectores. SQL no fue diseñado para búsquedas vectoriales intensivas, como su oponente, las búsquedas vectoriales.
